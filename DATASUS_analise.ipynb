```python
from google.colab import files
import io
import pandas as pd
import pygwalker as pyg

# 1. Faz o upload do arquivo no Colab
uploaded = files.upload()

# 2. Pega o nome do primeiro arquivo enviado
nome_arquivo = list(uploaded.keys())[0]

# 3. Lê o CSV usando o conteúdo em memória e o separador ';'
df = pd.read_csv(io.BytesIO(uploaded[nome_arquivo]), sep=";", header=0)

# 4. Abre a interface interativa do Pygwalker
pyg.walk(df)
```
